```sql
-- 1. null values 
select  count(*) as total_records,
        sum(case when day is null then 1 else 0 end) as missing_day,
        sum(case when environment is null or trim(environment) = '' then 1 else 0 end) as missing_environment,
        sum(case when device is null or trim(device) = '' then 1 else 0 end) as missing_device,
        sum(case when medium is null or trim(medium) = '' then 1 else 0 end) as missing_medium,
        sum(case when business_market is null or trim(business_market) = '' then 1 else 0 end) as missing_market,
        sum(case when sessions is null then 1 else 0 end) as missing_sessions
from workspace.redfin_bronze.redfin_user_growth
```

```sql
--1a. null environment 
select
    case  when sessions = 0 then 'zero sessions'
          when sessions > 0 then 'positive sessions'
         when sessions < 0 then 'negative sessions'
    end as session_status,
    count(*) as record_count,
    sum(sessions) as total_sessions
from workspace.redfin_bronze.redfin_user_growth
where environment is null or trim(environment) = ''
group by 1  order by 1 
```

```sql
-- 2 duplicates  
select
    day,
    environment,
    device,
    medium,
    business_market,
    count(*) as record_count
from workspace.redfin_bronze.redfin_user_growth
group by   1,2,3,4,5 
having count(*) > 1
order by record_count desc
```

```sql
-- 3. valid sessions 
select
    count(*) as total_records,
    sum(case when sessions < 0 then 1 else 0 end) as negative_sessions,
    sum(case when sessions = 0 then 1 else 0 end) as zero_sessions,
    sum(case when sessions > 0 then 1 else 0 end) as positive_sessions
from workspace.redfin_bronze.redfin_user_growth 
```

```sql
--3.1 
select distinct environment , count(*)
from workspace.redfin_bronze.redfin_user_growth 
where sessions = 0 
group by 1 
```

```sql
--3.2 
select   environment,
        count(*) as record_count,
        count(distinct day) as observed_days,
        sum(sessions) as total_sessions
from workspace.redfin_bronze.redfin_user_growth
where sessions = 0
--  and environment is not null
--  and trim(environment) <> ''
group by 1 
```

```sql
--4. dimensions 
select
    'environment' as dimension,
    coalesce(nullif(trim(environment), ''), '<missing>') as dimension_value,
    count(*) as record_count
from workspace.redfin_bronze.redfin_user_growth
group by coalesce(nullif(trim(environment), ''), '<missing>')

union all

select
    'device' as dimension,
    coalesce(nullif(trim(device), ''), '<missing>') as dimension_value,
    count(*) as record_count
from workspace.redfin_bronze.redfin_user_growth
group by coalesce(nullif(trim(device), ''), '<missing>')

union all

select
    'medium' as dimension,
    coalesce(nullif(trim(medium), ''), '<missing>') as dimension_value,
    count(*) as record_count
from workspace.redfin_bronze.redfin_user_growth
group by coalesce(nullif(trim(medium), ''), '<missing>')

order by dimension, dimension_value;
```

```sql
select
    coalesce(nullif(trim(business_market), ''), '<missing>') as business_market,
    count(*) as record_count,
    sum(sessions) as total_sessions
from workspace.redfin_bronze.redfin_user_growth
group by coalesce(nullif(trim(business_market), ''), '<missing>')
order by record_count desc;
```

```sql
select
    'overall' as grouping,
    sum(sessions) as total_sessions
from workspace.redfin_bronze.redfin_user_growth

union all

select
    'environment' as grouping,
    sum(sessions) as total_sessions
from workspace.redfin_bronze.redfin_user_growth
group by environment

union all

select
    'device' as grouping,
    sum(sessions) as total_sessions
from workspace.redfin_bronze.redfin_user_growth
group by device

union all

select
    'medium' as grouping,
    sum(sessions) as total_sessions
from workspace.redfin_bronze.redfin_user_growth
group by medium

union all

select
    'business_market' as grouping,
    sum(sessions) as total_sessions
from workspace.redfin_bronze.redfin_user_growth
group by business_market;
```

```sql
select
    'environment' as dimension,
    count(*) as affected_records
from workspace.redfin_bronze.redfin_user_growth
where environment is not null
  and environment <> trim(environment)

union all

select
    'device' as dimension,
    count(*) as affected_records
from workspace.redfin_bronze.redfin_user_growth
where device is not null
  and device <> trim(device)

union all

select
    'medium' as dimension,
    count(*) as affected_records
from workspace.redfin_bronze.redfin_user_growth
where medium is not null
  and medium <> trim(medium)

union all

select
    'business_market' as dimension,
    count(*) as affected_records
from workspace.redfin_bronze.redfin_user_growth
where business_market is not null
  and business_market <> trim(business_market);
```
